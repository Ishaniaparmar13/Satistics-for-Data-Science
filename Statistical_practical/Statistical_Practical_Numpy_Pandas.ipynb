```python
print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("\nDataset Information:")
df.info()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
numeric_data = df[
    ["Age", "Study_Hours", "Marks"]
]

X = numeric_data.to_numpy()

print(X)
```

```python
print("Shape:", X.shape)
print("Size:", X.size)
print("Number of dimensions:", X.ndim)
```

```python
A = X[:4, :]
B = X[4:8, :]

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
addition = A + B

print("A + B =")
print(addition)
```

```python
subtraction = A - B

print("A - B =")
print(subtraction)
```

```python
multiplication = A * B

print("Element-wise multiplication A * B =")
print(multiplication)
```

```python
matrix_product = np.dot(A, B.T)

print("Matrix product A × Bᵀ =")
print(matrix_product)
```

```python
transpose_A = A.T

print("Transpose of A:")
print(transpose_A)
```

```python
print("Original shape:", A.shape)
print("Transpose shape:", A.T.shape)
```

```python
M = X[:3, :3]

print("Matrix M:")
print(M)

print("\nShape of M:")
print(M.shape)

print("\nDeterminant of M:")
print(np.linalg.det(M))
```

```python
M_inverse = np.linalg.inv(M)

print("Inverse of M:")
print(M_inverse)
```

```python
identity_check = np.dot(M, M_inverse)

print("M × M_inverse =")
print(identity_check)
```
